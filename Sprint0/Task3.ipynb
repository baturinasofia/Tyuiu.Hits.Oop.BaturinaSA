```
using System;
using System.Collections.Generic;

public class Product
{
    public string Name { get; set; }
    public decimal Price { get; set; }
    public string Manufacturer { get; set; }

    public Product(string name, decimal price, string manufacturer)
    {
        Name = name;
        Price = price;
        Manufacturer = manufacturer;
    }

    public virtual string GetInfo()
    {
        return $"Название: {Name}, Производитель: {Manufacturer}, Цена: {Price:C}";
    }

    public virtual decimal Discount()
    {
        decimal discountPrice = Price * 0.9m; // Скидка 10%
        Console.WriteLine($"[Product] Применена скидка 10% к '{Name}'. Новая цена: {discountPrice:C}");
        return discountPrice;
    }

    public virtual void Display()
    {
        Console.WriteLine($"=== {Name} ===");
        Console.WriteLine($"Производитель: {Manufacturer}");
        Console.WriteLine($"Цена: {Price:C}");
    }
}

public class Electronics : Product
{
    public int WarrantyPeriod { get; set; } // в месяцах

    public Electronics(string name, decimal price, string manufacturer, int warrantyPeriod)
        : base(name, price, manufacturer)
    {
        WarrantyPeriod = warrantyPeriod;
    }

    public override decimal Discount()
    {
        // Чем больше гарантия, тем меньше скидка
        decimal discountRate = WarrantyPeriod >= 24 ? 0.95m : 0.85m;
        decimal discountPrice = Price * discountRate;
        
        Console.WriteLine($"[Electronics] Скидка для '{Name}' (гарантия {WarrantyPeriod} мес.): " +
                         $"{(100 - discountRate * 100)}%. Новая цена: {discountPrice:C}");
        return discountPrice;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Гарантия: {WarrantyPeriod} мес.";
    }

    public void ExtendWarranty(int extraMonths)
    {
        WarrantyPeriod += extraMonths;
        Console.WriteLine($"[Electronics] Гарантия для '{Name}' продлена на {extraMonths} мес. " +
                         $"Итого: {WarrantyPeriod} мес.");
    }
}

public class Clothing : Product
{
    public string Size { get; set; }

    public Clothing(string name, decimal price, string manufacturer, string size)
        : base(name, price, manufacturer)
    {
        Size = size;
    }

    public override void Display()
    {
        Console.WriteLine($"=== {Name} ===");
        Console.WriteLine($"Производитель: {Manufacturer}");
        Console.WriteLine($"Размер: {Size}");
        Console.WriteLine($"Цена: {Price:C}");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Размер: {Size}";
    }

    public bool IsSizeAvailable(string customerSize)
    {
        bool isAvailable = customerSize == Size;
        Console.WriteLine($"[Clothing] Размер '{customerSize}' для '{Name}': " +
                         $"{(isAvailable ? "в наличии" : "нет в наличии")}");
        return isAvailable;
    }
}

public class Books : Product
{
    public string Author { get; set; }
    public int Pages { get; set; }

    public Books(string name, decimal price, string manufacturer, string author, int pages)
        : base(name, price, manufacturer)
    {
        Author = author;
        Pages = pages;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Автор: {Author}, Страниц: {Pages}";
    }

    public override decimal Discount()
    {
        // Скидка зависит от количества страниц
        decimal discountRate = Pages > 500 ? 0.8m : 0.9m;
        decimal discountPrice = Price * discountRate;
        
        Console.WriteLine($"[Books] Скидка для книги '{Name}' ({Pages} стр.): " +
                         $"{(100 - discountRate * 100)}%. Новая цена: {discountPrice:C}");
        return discountPrice;
    }

    public void ShowAuthorInfo()
    {
        Console.WriteLine($"[Books] Книга '{Name}' написана автором: {Author}");
    }
}

// ==================== ОСНОВНАЯ ПРОГРАММА ====================

var products = new List<Product>
{
    new Electronics("Смартфон Galaxy S23", 79999m, "Samsung", 24),
    new Electronics("Ноутбук ProBook", 125000m, "HP", 12),
    new Clothing("Футболка спортивная", 2499m, "Adidas", "L"),
    new Clothing("Джинсы классические", 4999m, "Levi's", "32"),
    new Books("Изучаем C#", 2999m, "Питер", "Герберт Шилдт", 1200),
    new Books("Алгоритмы", 3499m, "Вильямс", "Томас Кормен", 960)
};

Console.WriteLine("=== ИНФОРМАЦИЯ О ВСЕХ ПРОДУКТАХ ===\n");
foreach (var p in products)
{
    Console.WriteLine($"• {p.GetInfo()}");
}

Console.WriteLine("\n=== ОТОБРАЖЕНИЕ ПРОДУКТОВ ===\n");
foreach (var p in products)
{
    p.Display();
    Console.WriteLine();
}

Console.WriteLine("=== ПРИМЕНЕНИЕ СКИДОК ===\n");
foreach (var p in products)
{
    p.Discount();
}

Console.WriteLine("\n=== ДОПОЛНИТЕЛЬНЫЕ МЕТОДЫ ===\n");

// Работа с электроникой
var electronics = (Electronics)products[0];
electronics.ExtendWarranty(12);

// Работа с одеждой
var clothing = (Clothing)products[2];
clothing.IsSizeAvailable("L");
clothing.IsSizeAvailable("M");

// Работа с книгами
var book = (Books)products[4];
book.ShowAuthorInfo();

Console.WriteLine("\n=== ИТОГОВАЯ ИНФОРМАЦИЯ ===\n");
foreach (var p in products)
{
    Console.WriteLine($"• {p.GetInfo()}");
}
```
