```
using System;
using System.Collections.Generic;

// ==================== ИНТЕРФЕЙСЫ ДЛЯ МНОЖЕСТВЕННОГО НАСЛЕДОВАНИЯ ====================

public interface IRechargeable
{
    void Recharge();
    int BatteryLevel { get; set; }
}

public interface IWearable
{
    void Wear();
    string Material { get; }
}

public interface IReadable
{
    void Read();
    string Language { get; set; }
}

// ==================== БАЗОВЫЙ КЛАСС PRODUCT ====================

public class Product
{
    // Приватные поля для инкапсуляции
    private string _name;
    private decimal _price;
    private string _manufacturer;
    private string _category;      // НОВЫЙ атрибут
    private string _country;       // НОВЫЙ атрибут
    private double _rating;        // НОВЫЙ атрибут

    // Свойства с явными геттерами и сеттерами (с валидацией)
    public string Name
    {
        get { return _name; }
        set
        {
            if (!string.IsNullOrWhiteSpace(value))
                _name = value;
            else
                throw new ArgumentException("Название продукта не может быть пустым!");
        }
    }

    public decimal Price
    {
        get { return _price; }
        set
        {
            if (value > 0)
                _price = value;
            else
                throw new ArgumentOutOfRangeException("Цена должна быть больше нуля!");
        }
    }

    public string Manufacturer
    {
        get { return _manufacturer; }
        set
        {
            if (!string.IsNullOrWhiteSpace(value))
                _manufacturer = value;
            else
                throw new ArgumentException("Производитель не может быть пустым!");
        }
    }

    // НОВЫЕ свойства
    public string Category
    {
        get { return _category; }
        set { _category = value ?? "Без категории"; }
    }

    public string Country
    {
        get { return _country; }
        set { _country = value ?? "Неизвестно"; }
    }

    public double Rating
    {
        get { return _rating; }
        set
        {
            if (value >= 0 && value <= 5)
                _rating = value;
            else
                throw new ArgumentOutOfRangeException("Рейтинг должен быть от 0 до 5!");
        }
    }

    // Конструктор с использованием сеттеров
    public Product(string name, decimal price, string manufacturer, string category, string country, double rating)
    {
        Name = name;
        Price = price;
        Manufacturer = manufacturer;
        Category = category;
        Country = country;
        Rating = rating;
    }

    public virtual string GetInfo()
    {
        return $"Название: {Name}, Производитель: {Manufacturer}, Цена: {Price:C}, " +
               $"Категория: {Category}, Страна: {Country}, Рейтинг: {Rating}";
    }

    public virtual decimal Discount()
    {
        decimal discountPrice = Price * 0.9m;
        Console.WriteLine($"[Product] Применена скидка 10% к '{Name}'. Новая цена: {discountPrice:C}");
        return discountPrice;
    }

    public virtual void Display()
    {
        Console.WriteLine($"=== {Name} ===");
        Console.WriteLine($"Производитель: {Manufacturer}");
        Console.WriteLine($"Категория: {Category}");
        Console.WriteLine($"Страна: {Country}");
        Console.WriteLine($"Рейтинг: {Rating}");
        Console.WriteLine($"Цена: {Price:C}");
    }

    // НОВЫЕ методы
    public void UpdatePrice(decimal newPrice)
    {
        Price = newPrice;
        Console.WriteLine($"[Product] Цена '{Name}' обновлена: {Price:C}");
    }

    public string GetCategory()
    {
        return $"Категория продукта '{Name}': {Category}";
    }

    public void IncreaseRating(double points)
    {
        double newRating = Rating + points;
        if (newRating <= 5)
        {
            Rating = newRating;
            Console.WriteLine($"[Product] Рейтинг '{Name}' увеличен на {points}. Новый рейтинг: {Rating}");
        }
        else
        {
            Console.WriteLine($"[Product] Рейтинг '{Name}' уже максимален (5.0)");
        }
    }
}

// ==================== ПРОИЗВОДНЫЙ КЛАСС ELECTRONICS (ПРОСТОЕ НАСЛЕДОВАНИЕ) ====================

public class Electronics : Product
{
    private int _warrantyPeriod;
    private int _power;     // НОВЫЙ атрибут
    private string _color;  // НОВЫЙ атрибут

    public int WarrantyPeriod
    {
        get { return _warrantyPeriod; }
        set
        {
            if (value >= 0)
                _warrantyPeriod = value;
            else
                throw new ArgumentOutOfRangeException("Гарантийный срок не может быть отрицательным!");
        }
    }

    // НОВЫЕ свойства
    public int Power
    {
        get { return _power; }
        set
        {
            if (value > 0)
                _power = value;
            else
                throw new ArgumentOutOfRangeException("Мощность должна быть больше нуля!");
        }
    }

    public string Color
    {
        get { return _color; }
        set { _color = value ?? "Черный"; }
    }

    public Electronics(string name, decimal price, string manufacturer, int warrantyPeriod, 
                       string category, string country, double rating, int power, string color)
        : base(name, price, manufacturer, category, country, rating)
    {
        WarrantyPeriod = warrantyPeriod;
        Power = power;
        Color = color;
    }

    public override decimal Discount()
    {
        decimal discountRate = WarrantyPeriod >= 24 ? 0.95m : 0.85m;
        decimal discountPrice = Price * discountRate;
        
        Console.WriteLine($"[Electronics] Скидка для '{Name}' (гарантия {WarrantyPeriod} мес.): " +
                         $"{(100 - discountRate * 100)}%. Новая цена: {discountPrice:C}");
        return discountPrice;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Гарантия: {WarrantyPeriod} мес., Мощность: {Power} Вт, Цвет: {Color}";
    }

    public override void Display()
    {
        base.Display();
        Console.WriteLine($"Гарантия: {WarrantyPeriod} мес.");
        Console.WriteLine($"Мощность: {Power} Вт");
        Console.WriteLine($"Цвет: {Color}");
    }

    public void ExtendWarranty(int extraMonths)
    {
        WarrantyPeriod += extraMonths;
        Console.WriteLine($"[Electronics] Гарантия для '{Name}' продлена на {extraMonths} мес. " +
                         $"Итого: {WarrantyPeriod} мес.");
    }

    // НОВЫЕ методы
    public void TurnOn()
    {
        Console.WriteLine($"[Electronics] Устройство '{Name}' включено. Потребление: {Power} Вт");
    }

    public string GetPowerInfo()
    {
        return $"Мощность '{Name}': {Power} Вт";
    }
}

// ==================== СЛОЖНОЕ НАСЛЕДОВАНИЕ: SMARTPHONE НАСЛЕДУЕТСЯ ОТ ELECTRONICS ====================

public class Smartphone : Electronics, IRechargeable
{
    private string _os;           // Операционная система
    private int _batteryLevel;    // Уровень батареи
    private int _storageGb;       // Объем памяти

    public string OS
    {
        get { return _os; }
        set { _os = value ?? "Android"; }
    }

    public int BatteryLevel
    {
        get { return _batteryLevel; }
        set
        {
            if (value >= 0 && value <= 100)
                _batteryLevel = value;
            else
                throw new ArgumentOutOfRangeException("Уровень батареи должен быть от 0 до 100!");
        }
    }

    public int StorageGb
    {
        get { return _storageGb; }
        set
        {
            if (value > 0)
                _storageGb = value;
            else
                throw new ArgumentOutOfRangeException("Объем памяти должен быть больше нуля!");
        }
    }

    public Smartphone(string name, decimal price, string manufacturer, int warrantyPeriod,
                      string category, string country, double rating, int power, string color,
                      string os, int batteryLevel, int storageGb)
        : base(name, price, manufacturer, warrantyPeriod, category, country, rating, power, color)
    {
        OS = os;
        BatteryLevel = batteryLevel;
        StorageGb = storageGb;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", ОС: {OS}, Память: {StorageGb} ГБ, Батарея: {BatteryLevel}%";
    }

    public override void Display()
    {
        base.Display();
        Console.WriteLine($"ОС: {OS}");
        Console.WriteLine($"Память: {StorageGb} ГБ");
        Console.WriteLine($"Батарея: {BatteryLevel}%");
    }

    // Реализация интерфейса IRechargeable
    public void Recharge()
    {
        BatteryLevel = 100;
        Console.WriteLine($"[Smartphone] '{Name}' полностью заряжен. Батарея: {BatteryLevel}%");
    }

    public void MakeCall(string number)
    {
        Console.WriteLine($"[Smartphone] '{Name}' звонит на номер {number}");
    }

    public void InstallApp(string appName)
    {
        Console.WriteLine($"[Smartphone] На '{Name}' установлено приложение '{appName}'");
    }
}

// ==================== ПРОИЗВОДНЫЙ КЛАСС CLOTHING (ПРОСТОЕ НАСЛЕДОВАНИЕ) ====================

public class Clothing : Product
{
    private string _size;
    private string _material;   // НОВЫЙ атрибут
    private string _season;     // НОВЫЙ атрибут

    public string Size
    {
        get { return _size; }
        set
        {
            if (!string.IsNullOrWhiteSpace(value))
                _size = value;
            else
                throw new ArgumentException("Размер не может быть пустым!");
        }
    }

    // НОВЫЕ свойства
    public string Material
    {
        get { return _material; }
        set { _material = value ?? "Хлопок"; }
    }

    public string Season
    {
        get { return _season; }
        set { _season = value ?? "Всесезонный"; }
    }

    public Clothing(string name, decimal price, string manufacturer, string size,
                    string category, string country, double rating, string material, string season)
        : base(name, price, manufacturer, category, country, rating)
    {
        Size = size;
        Material = material;
        Season = season;
    }

    public override void Display()
    {
        base.Display();
        Console.WriteLine($"Размер: {Size}");
        Console.WriteLine($"Материал: {Material}");
        Console.WriteLine($"Сезон: {Season}");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Размер: {Size}, Материал: {Material}, Сезон: {Season}";
    }

    public bool IsSizeAvailable(string customerSize)
    {
        bool isAvailable = customerSize == Size;
        Console.WriteLine($"[Clothing] Размер '{customerSize}' для '{Name}': " +
                         $"{(isAvailable ? "в наличии" : "нет в наличии")}");
        return isAvailable;
    }

    // НОВЫЕ методы
    public string GetSeasonInfo()
    {
        return $"Одежда '{Name}' предназначена для сезона: {Season}";
    }

    public void ChangeSize(string newSize)
    {
        Size = newSize;
        Console.WriteLine($"[Clothing] Размер '{Name}' изменен на {Size}");
    }
}

// ==================== МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ: SMARTWATCH НАСЛЕДУЕТ ELECTRONICS И РЕАЛИЗУЕТ IWearable ====================

public class SmartWatch : Electronics, IWearable
{
    private string _strapMaterial;
    private bool _isWaterproof;

    public string StrapMaterial
    {
        get { return _strapMaterial; }
        set { _strapMaterial = value ?? "Силикон"; }
    }

    public bool IsWaterproof
    {
        get { return _isWaterproof; }
        set { _isWaterproof = value; }
    }

    // Реализация интерфейса IWearable
    string IWearable.Material => StrapMaterial;

    public SmartWatch(string name, decimal price, string manufacturer, int warrantyPeriod,
                      string category, string country, double rating, int power, string color,
                      string strapMaterial, bool isWaterproof)
        : base(name, price, manufacturer, warrantyPeriod, category, country, rating, power, color)
    {
        StrapMaterial = strapMaterial;
        IsWaterproof = isWaterproof;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Ремешок: {StrapMaterial}, Водонепроницаемый: {(IsWaterproof ? "Да" : "Нет")}";
    }

    public override void Display()
    {
        base.Display();
        Console.WriteLine($"Ремешок: {StrapMaterial}");
        Console.WriteLine($"Водонепроницаемый: {(IsWaterproof ? "Да" : "Нет")}");
    }

    // Реализация интерфейса IWearable
    public void Wear()
    {
        Console.WriteLine($"[SmartWatch] '{Name}' надет на запястье. Материал ремешка: {StrapMaterial}");
    }

    public void TrackHeartRate()
    {
        Console.WriteLine($"[SmartWatch] '{Name}' измеряет пульс: 72 уд/мин");
    }

    public void ShowNotifications()
    {
        Console.WriteLine($"[SmartWatch] '{Name}' показывает 3 новых уведомления");
    }
}

// ==================== ПРОИЗВОДНЫЙ КЛАСС BOOKS (ПРОСТОЕ НАСЛЕДОВАНИЕ) ====================

public class Books : Product, IReadable
{
    private string _author;
    private int _pages;
    private string _genre;      // НОВЫЙ атрибут
    private string _language;   // НОВЫЙ атрибут

    public string Author
    {
        get { return _author; }
        set
        {
            if (!string.IsNullOrWhiteSpace(value))
                _author = value;
            else
                throw new ArgumentException("Автор не может быть пустым!");
        }
    }

    public int Pages
    {
        get { return _pages; }
        set
        {
            if (value > 0)
                _pages = value;
            else
                throw new ArgumentOutOfRangeException("Количество страниц должно быть больше нуля!");
        }
    }

    // НОВЫЕ свойства
    public string Genre
    {
        get { return _genre; }
        set { _genre = value ?? "Не указан"; }
    }

    public string Language
    {
        get { return _language; }
        set { _language = value ?? "Русский"; }
    }

    public Books(string name, decimal price, string manufacturer, string author, int pages,
                 string category, string country, double rating, string genre, string language)
        : base(name, price, manufacturer, category, country, rating)
    {
        Author = author;
        Pages = pages;
        Genre = genre;
        Language = language;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Автор: {Author}, Страниц: {Pages}, Жанр: {Genre}, Язык: {Language}";
    }

    public override decimal Discount()
    {
        decimal discountRate = Pages > 500 ? 0.8m : 0.9m;
        decimal discountPrice = Price * discountRate;
        
        Console.WriteLine($"[Books] Скидка для книги '{Name}' ({Pages} стр.): " +
                         $"{(100 - discountRate * 100)}%. Новая цена: {discountPrice:C}");
        return discountPrice;
    }

    public void ShowAuthorInfo()
    {
        Console.WriteLine($"[Books] Книга '{Name}' написана автором: {Author}");
    }

    // Реализация интерфейса IReadable
    public void Read()
    {
        Console.WriteLine($"[Books] Чтение книги '{Name}' на языке {Language}");
    }

    // НОВЫЕ методы
    public string GetGenreInfo()
    {
        return $"Жанр книги '{Name}': {Genre}";
    }

    public void Translate(string newLanguage)
    {
        Language = newLanguage;
        Console.WriteLine($"[Books] Книга '{Name}' переведена на язык: {Language}");
    }
}

// ==================== СЛОЖНОЕ НАСЛЕДОВАНИЕ: TEXTBOOK НАСЛЕДУЕТСЯ ОТ BOOKS ====================

public class Textbook : Books
{
    private string _subject;
    private int _grade;

    public string Subject
    {
        get { return _subject; }
        set { _subject = value ?? "Общий"; }
    }

    public int Grade
    {
        get { return _grade; }
        set
        {
            if (value >= 1 && value <= 11)
                _grade = value;
            else
                throw new ArgumentOutOfRangeException("Класс должен быть от 1 до 11!");
        }
    }

    public Textbook(string name, decimal price, string manufacturer, string author, int pages,
                    string category, string country, double rating, string genre, string language,
                    string subject, int grade)
        : base(name, price, manufacturer, author, pages, category, country, rating, genre, language)
    {
        Subject = subject;
        Grade = grade;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Предмет: {Subject}, Класс: {Grade}";
    }

    public override void Display()
    {
        base.Display();
        Console.WriteLine($"Предмет: {Subject}");
        Console.WriteLine($"Класс: {Grade}");
    }

    public void GetHomework()
    {
        Console.WriteLine($"[Textbook] Учебник '{Name}' ({Subject}, {Grade} класс): выполните упражнение 5");
    }
}

// ==================== ВЗАИМОДЕЙСТВИЕ ОБЪЕКТОВ ====================

public class ShoppingCart
{
    private List<Product> _items;
    private string _customerName;

    public string CustomerName
    {
        get { return _customerName; }
        set { _customerName = value; }
    }

    public ShoppingCart(string customerName)
    {
        _items = new List<Product>();
        CustomerName = customerName;
    }

    public void AddProduct(Product product)
    {
        _items.Add(product);
        Console.WriteLine($"[Корзина] {product.Name} добавлен в корзину {CustomerName}");
    }

    public void ApplyDiscountsToAll()
    {
        Console.WriteLine($"\n[Корзина] Применение скидок для {CustomerName}:");
        foreach (var item in _items)
        {
            item.Discount();
        }
    }

    public decimal CalculateTotal()
    {
        decimal total = 0;
        foreach (var item in _items)
        {
            total += item.Price;
        }
        return total;
    }

    public void DisplayCart()
    {
        Console.WriteLine($"\n=== Корзина покупателя: {CustomerName} ===");
        foreach (var item in _items)
        {
            Console.WriteLine($"• {item.GetInfo()}");
        }
        Console.WriteLine($"\nОбщая стоимость: {CalculateTotal():C}");
    }

    public void FindByManufacturer(string manufacturer)
    {
        Console.WriteLine($"\n[Корзина] Товары от производителя '{manufacturer}':");
        bool found = false;
        foreach (var item in _items)
        {
            if (item.Manufacturer.Equals(manufacturer, StringComparison.OrdinalIgnoreCase))
            {
                Console.WriteLine($"  ✓ {item.Name}");
                found = true;
            }
        }
        if (!found)
            Console.WriteLine("  Товары не найдены");
    }

    // НОВЫЙ метод: фильтрация по категории
    public void FilterByCategory(string category)
    {
        Console.WriteLine($"\n[Корзина] Товары категории '{category}':");
        foreach (var item in _items)
        {
            if (item.Category.Equals(category, StringComparison.OrdinalIgnoreCase))
            {
                Console.WriteLine($"  ✓ {item.Name} - {item.Price:C}");
            }
        }
    }
}

// ==================== ОСНОВНАЯ ПРОГРАММА ====================

Console.WriteLine("=== СОЗДАНИЕ ПРОДУКТОВ ===\n");

var products = new List<Product>
{
    new Electronics("Смартфон Galaxy S23", 79999m, "Samsung", 24, "Электроника", "Южная Корея", 4.8, 15, "Черный"),
    new Smartphone("iPhone 15 Pro", 129999m, "Apple", 12, "Смартфоны", "США", 4.9, 10, "Титановый", "iOS", 85, 256),
    new SmartWatch("Apple Watch Ultra", 89999m, "Apple", 12, "Часы", "США", 4.7, 5, "Оранжевый", "Фторкаучук", true),
    new Clothing("Футболка спортивная", 2499m, "Adidas", "L", "Одежда", "Германия", 4.5, "Полиэстер", "Лето"),
    new Clothing("Джинсы классические", 4999m, "Levi's", "32", "Одежда", "США", 4.6, "Деним", "Всесезонный"),
    new Books("Изучаем C#", 2999m, "Питер", "Герберт Шилдт", 1200, "Книги", "США", 4.7, "Техническая", "Русский"),
    new Textbook("Алгоритмы для школьников", 1499m, "Просвещение", "Томас Кормен", 320, "Учебники", "Россия", 4.4, "Образовательная", "Русский", "Информатика", 10)
};

Console.WriteLine("\n=== ИНФОРМАЦИЯ О ВСЕХ ПРОДУКТАХ ===\n");
foreach (var p in products)
{
    Console.WriteLine($"• {p.GetInfo()}");
}

Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ПРОСТОГО НАСЛЕДОВАНИЯ ===\n");
Console.WriteLine("[Electronics наследует Product]");
products[0].Display();

Console.WriteLine("\n[Clothing наследует Product]");
products[3].Display();

Console.WriteLine("\n[Books наследует Product]");
products[5].Display();

Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ СЛОЖНОГО НАСЛЕДОВАНИЯ ===\n");
Console.WriteLine("[Smartphone наследует Electronics, который наследует Product]");
var smartphone = (Smartphone)products[1];
smartphone.Display();
smartphone.MakeCall("+7-999-123-45-67");
smartphone.Recharge();

Console.WriteLine("\n[Textbook наследует Books, который наследует Product]");
var textbook = (Textbook)products[6];
textbook.Display();
textbook.GetHomework();

Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ МНОЖЕСТВЕННОГО НАСЛЕДОВАНИЯ ===\n");
Console.WriteLine("[SmartWatch наследует Electronics и реализует интерфейс IWearable]");
var smartWatch = (SmartWatch)products[2];
smartWatch.Display();
smartWatch.Wear();  // Метод из интерфейса IWearable
smartWatch.TrackHeartRate();
smartWatch.ShowNotifications();

Console.WriteLine("\n=== ВЗАИМОДЕЙСТВИЕ ОБЪЕКТОВ: КОРЗИНА ПОКУПАТЕЛЯ ===\n");

var cart1 = new ShoppingCart("Иван Петров");
cart1.AddProduct(products[0]);
cart1.AddProduct(products[3]);
cart1.AddProduct(products[5]);

cart1.DisplayCart();
cart1.ApplyDiscountsToAll();
cart1.FindByManufacturer("Samsung");
cart1.FilterByCategory("Электроника");

var cart2 = new ShoppingCart("Мария Сидорова");
cart2.AddProduct(products[1]);
cart2.AddProduct(products[4]);
cart2.AddProduct(products[6]);

cart2.DisplayCart();
cart2.ApplyDiscountsToAll();

Console.WriteLine("\n=== ДОПОЛНИТЕЛЬНЫЕ МЕТОДЫ ===\n");

var electronics = (Electronics)products[0];
electronics.ExtendWarranty(12);
electronics.TurnOn();
electronics.UpdatePrice(74999m);

var clothing = (Clothing)products[3];
clothing.IsSizeAvailable("L");
clothing.ChangeSize("XL");
Console.WriteLine(clothing.GetSeasonInfo());

var book = (Books)products[5];
book.ShowAuthorInfo();
book.Read();  // Метод из интерфейса IReadable
book.Translate("Английский");
Console.WriteLine(book.GetGenreInfo());
book.IncreaseRating(0.2);

Console.WriteLine("\n=== ИТОГОВАЯ ИНФОРМАЦИЯ ===\n");
foreach (var p in products)
{
    Console.WriteLine($"• {p.GetInfo()}");
}
```
