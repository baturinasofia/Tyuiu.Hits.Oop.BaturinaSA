```
using System;
using System.Collections.Generic;

public class Product
{
    protected string _name;
    protected decimal _price;
    protected string _manufacturer;

    public string Name
    {
        get { return _name; }
        set { _name = !string.IsNullOrWhiteSpace(value) ? value : throw new ArgumentException("Название не может быть пустым!"); }
    }

    public decimal Price
    {
        get { return _price; }
        set { _price = value > 0 ? value : throw new ArgumentOutOfRangeException("Цена должна быть больше нуля!"); }
    }

    public string Manufacturer
    {
        get { return _manufacturer; }
        set { _manufacturer = !string.IsNullOrWhiteSpace(value) ? value : throw new ArgumentException("Производитель не может быть пустым!"); }
    }

    public Product(string name, decimal price, string manufacturer)
    {
        Name = name;
        Price = price;
        Manufacturer = manufacturer;
    }

    public virtual string GetInfo()
    {
        return $"Название: {Name}, Производитель: {Manufacturer}, Цена: {Price:C}";
    }

    public virtual decimal Discount()
    {
        decimal discountPrice = Price * 0.9m;
        Console.WriteLine($"[Product] Применена скидка 10% к '{Name}'. Новая цена: {discountPrice:C}");
        return discountPrice;
    }

    public virtual void Display()
    {
        Console.WriteLine($"=== {Name} ===");
        Console.WriteLine($"Производитель: {Manufacturer}");
        Console.WriteLine($"Цена: {Price:C}");
    }
}

public class Electronics : Product
{
    private int _warrantyPeriod;

    public int WarrantyPeriod
    {
        get { return _warrantyPeriod; }
        set { _warrantyPeriod = value >= 0 ? value : throw new ArgumentOutOfRangeException("Гарантийный срок не может быть отрицательным!"); }
    }

    public Electronics(string name, decimal price, string manufacturer, int warrantyPeriod)
        : base(name, price, manufacturer)
    {
        WarrantyPeriod = warrantyPeriod;
    }

    public override decimal Discount()
    {
        decimal discountRate = WarrantyPeriod >= 24 ? 0.85m : 0.9m;
        decimal discountPrice = Price * discountRate;
        Console.WriteLine($"[Electronics] Скидка для '{Name}' (гарантия {WarrantyPeriod} мес.): {(100 - discountRate * 100)}%. Новая цена: {discountPrice:C}");
        return discountPrice;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Гарантия: {WarrantyPeriod} мес.";
    }

    public override void Display()
    {
        base.Display();
        Console.WriteLine($"Гарантия: {WarrantyPeriod} мес.");
    }
}

public class Clothing : Product
{
    private string _size;

    public string Size
    {
        get { return _size; }
        set { _size = !string.IsNullOrWhiteSpace(value) ? value : throw new ArgumentException("Размер не может быть пустым!"); }
    }

    public Clothing(string name, decimal price, string manufacturer, string size)
        : base(name, price, manufacturer)
    {
        Size = size;
    }

    public override void Display()
    {
        base.Display();
        Console.WriteLine($"Размер: {Size}");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Размер: {Size}";
    }
}

public class Books : Product
{
    private string _author;

    public string Author
    {
        get { return _author; }
        set { _author = !string.IsNullOrWhiteSpace(value) ? value : throw new ArgumentException("Автор не может быть пустым!"); }
    }

    public Books(string name, decimal price, string manufacturer, string author)
        : base(name, price, manufacturer)
    {
        Author = author;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Автор: {Author}";
    }

    public override void Display()
    {
        base.Display();
        Console.WriteLine($"Автор: {Author}");
    }
}

List<Product> products = new List<Product>
{
    new Electronics("Смартфон Galaxy S23", 79999m, "Samsung", 24),
    new Electronics("Наушники Sony WH-1000", 29999m, "Sony", 12),
    new Clothing("Футболка спортивная", 2499m, "Adidas", "L"),
    new Clothing("Джинсы классические", 4999m, "Levi's", "32"),
    new Books("Изучаем C#", 2999m, "Питер", "Герберт Шилдт"),
    new Books("Чистый код", 3499m, "Питер", "Роберт Мартин")
};

Console.WriteLine("=== ИНФОРМАЦИЯ О ВСЕХ ПРОДУКТАХ ===\n");
foreach (var p in products)
{
    Console.WriteLine($"• {p.GetInfo()}");
}

Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ПРОСТОГО НАСЛЕДОВАНИЯ (Display) ===\n");
Console.WriteLine("[Electronics]");
products[0].Display();
Console.WriteLine();

Console.WriteLine("[Clothing]");
products[2].Display();
Console.WriteLine();

Console.WriteLine("[Books]");
products[4].Display();

Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА (Discount) ===\n");
foreach (var p in products)
{
    p.Discount();
}

Console.WriteLine("\n=== ПРОВЕРКА ПЕРЕОПРЕДЕЛЕНИЯ МЕТОДОВ ===\n");
var electronics = (Electronics)products[0];
Console.WriteLine($"Electronics GetInfo: {electronics.GetInfo()}");
Console.WriteLine($"Electronics WarrantyPeriod: {electronics.WarrantyPeriod} мес.");

var clothing = (Clothing)products[2];
Console.WriteLine($"\nClothing GetInfo: {clothing.GetInfo()}");
Console.WriteLine($"Clothing Size: {clothing.Size}");

var books = (Books)products[4];
Console.WriteLine($"\nBooks GetInfo: {books.GetInfo()}");
Console.WriteLine($"Books Author: {books.Author}");
```
