```
using System;
using System.Collections.Generic;

// ==================== ИНТЕРФЕЙС С ЯВНОЙ РЕАЛИЗАЦИЕЙ ====================

public interface IProductActions
{
    void Restock(int quantity);
    string GetCategory();
    string GetWarrantyInfo();
}

// ==================== БАЗОВЫЙ КЛАСС PRODUCT ====================

public class Product : IProductActions
{
    private string _name;
    private decimal _price;
    private string _manufacturer;
    private string _category;
    private string _country;
    private double _rating;
    private int _stockQuantity;

    public string Name
    {
        get { return _name; }
        set { _name = !string.IsNullOrWhiteSpace(value) ? value : throw new ArgumentException("Название не может быть пустым!"); }
    }

    public decimal Price
    {
        get { return _price; }
        set { _price = value > 0 ? value : throw new ArgumentOutOfRangeException("Цена должна быть больше нуля!"); }
    }

    public string Manufacturer
    {
        get { return _manufacturer; }
        set { _manufacturer = !string.IsNullOrWhiteSpace(value) ? value : throw new ArgumentException("Производитель не может быть пустым!"); }
    }

    public string Category
    {
        get { return _category; }
        set { _category = value ?? "Без категории"; }
    }

    public string Country
    {
        get { return _country; }
        set { _country = value ?? "Неизвестно"; }
    }

    public double Rating
    {
        get { return _rating; }
        set { _rating = (value >= 0 && value <= 5) ? value : throw new ArgumentOutOfRangeException("Рейтинг от 0 до 5!"); }
    }

    public int StockQuantity
    {
        get { return _stockQuantity; }
        set { _stockQuantity = value >= 0 ? value : throw new ArgumentOutOfRangeException("Количество не может быть отрицательным!"); }
    }

    public Product(string name, decimal price, string manufacturer, string category, string country, double rating, int stockQuantity)
    {
        Name = name;
        Price = price;
        Manufacturer = manufacturer;
        Category = category;
        Country = country;
        Rating = rating;
        StockQuantity = stockQuantity;
    }

    public virtual string GetInfo()
    {
        return $"Название: {Name}, Производитель: {Manufacturer}, Цена: {Price:C}";
    }

    public virtual decimal Discount()
    {
        decimal discountPrice = Price * 0.9m;
        Console.WriteLine($"[Product] Скидка 10% для '{Name}'. Новая цена: {discountPrice:C}");
        return discountPrice;
    }

    public virtual void Display()
    {
        Console.WriteLine($"=== {Name} ===\nПроизводитель: {Manufacturer}\nКатегория: {Category}\nСтрана: {Country}\nРейтинг: {Rating}\nЦена: {Price:C}\nНа складе: {StockQuantity} шт.");
    }

    public void UpdatePrice(decimal newPrice)
    {
        Price = newPrice;
        Console.WriteLine($"[Product] Цена '{Name}' обновлена: {Price:C}");
    }

    public void IncreaseRating(double points)
    {
        double newRating = Rating + points;
        Rating = newRating <= 5 ? newRating : 5.0;
        Console.WriteLine($"[Product] Рейтинг '{Name}' увеличен до {Rating}");
    }

    public string GetStockInfo()
    {
        return $"Товар '{Name}': на складе {StockQuantity} шт.";
    }

    void IProductActions.Restock(int quantity)
    {
        StockQuantity += quantity;
        Console.WriteLine($"[IProductActions] '{Name}' пополнен на {quantity} шт. Всего: {StockQuantity} шт.");
    }

    string IProductActions.GetCategory()
    {
        return Category;
    }

    string IProductActions.GetWarrantyInfo()
    {
        return $"Гарантия для '{Name}': стандартная (1 год)";
    }
}

// ==================== ПРОИЗВОДНЫЙ КЛАСС ELECTRONICS ====================
// ИСПРАВЛЕНИЕ: добавлено ", IProductActions", чтобы разрешить явную реализацию в этом классе
public class Electronics : Product, IProductActions
{
    private int _warrantyPeriod;
    private int _power;
    private string _color;
    private bool _isWireless;

    public int WarrantyPeriod
    {
        get { return _warrantyPeriod; }
        set { _warrantyPeriod = value >= 0 ? value : throw new ArgumentOutOfRangeException("Гарантия не может быть отрицательной!"); }
    }

    public int Power
    {
        get { return _power; }
        set { _power = value > 0 ? value : throw new ArgumentOutOfRangeException("Мощность должна быть больше нуля!"); }
    }

    public string Color
    {
        get { return _color; }
        set { _color = value ?? "Черный"; }
    }

    public bool IsWireless
    {
        get { return _isWireless; }
        set { _isWireless = value; }
    }

    public Electronics(string name, decimal price, string manufacturer, int warrantyPeriod, 
                       string category, string country, double rating, int stockQuantity,
                       int power, string color, bool isWireless)
        : base(name, price, manufacturer, category, country, rating, stockQuantity)
    {
        WarrantyPeriod = warrantyPeriod;
        Power = power;
        Color = color;
        IsWireless = isWireless;
    }

    public override decimal Discount()
    {
        decimal rate = WarrantyPeriod >= 24 ? 0.85m : 0.9m;
        decimal discountPrice = Price * rate;
        Console.WriteLine($"[Electronics] Скидка для '{Name}' (гарантия {WarrantyPeriod} мес.): {(100 - rate * 100)}%. Цена: {discountPrice:C}");
        return discountPrice;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Гарантия: {WarrantyPeriod} мес., Мощность: {Power} Вт, Цвет: {Color}, Беспроводной: {(IsWireless ? "Да" : "Нет")}";
    }

    public override void Display()
    {
        base.Display();
        Console.WriteLine($"Гарантия: {WarrantyPeriod} мес.\nМощность: {Power} Вт\nЦвет: {Color}\nБеспроводной: {(IsWireless ? "Да" : "Нет")}");
    }

    public void TurnOn()
    {
        Console.WriteLine($"[Electronics] Устройство '{Name}' включено. Потребление: {Power} Вт");
    }

    public string GetPowerInfo()
    {
        return $"Мощность '{Name}': {Power} Вт";
    }

    public void ToggleWireless()
    {
        IsWireless = !IsWireless;
        Console.WriteLine($"[Electronics] '{Name}' теперь {(IsWireless ? "беспроводной" : "проводной")}");
    }

    // Теперь это работает, так как класс явно объявил реализацию IProductActions
    string IProductActions.GetWarrantyInfo()
    {
        return $"Гарантия для '{Name}': {WarrantyPeriod} месяцев";
    }
}

// ==================== ПРОИЗВОДНЫЙ КЛАСС CLOTHING ====================

public class Clothing : Product
{
    private string _size;
    private string _material;
    private string _season;
    private bool _isWashed;

    public string Size
    {
        get { return _size; }
        set { _size = !string.IsNullOrWhiteSpace(value) ? value : throw new ArgumentException("Размер не может быть пустым!"); }
    }

    public string Material
    {
        get { return _material; }
        set { _material = value ?? "Хлопок"; }
    }

    public string Season
    {
        get { return _season; }
        set { _season = value ?? "Всесезонный"; }
    }

    public bool IsWashed
    {
        get { return _isWashed; }
        set { _isWashed = value; }
    }

    public Clothing(string name, decimal price, string manufacturer, string size,
                    string category, string country, double rating, int stockQuantity,
                    string material, string season, bool isWashed)
        : base(name, price, manufacturer, category, country, rating, stockQuantity)
    {
        Size = size;
        Material = material;
        Season = season;
        IsWashed = isWashed;
    }

    public override void Display()
    {
        base.Display();
        Console.WriteLine($"Размер: {Size}\nМатериал: {Material}\nСезон: {Season}\nПостирано: {(IsWashed ? "Да" : "Нет")}");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Размер: {Size}, Материал: {Material}, Сезон: {Season}";
    }

    public void Wash()
    {
        IsWashed = true;
        Console.WriteLine($"[Clothing] '{Name}' постиран и готов к носке.");
    }

    public string GetSeasonInfo()
    {
        return $"Одежда '{Name}' предназначена для сезона: {Season}";
    }

    public void ChangeSize(string newSize)
    {
        Size = newSize;
        Console.WriteLine($"[Clothing] Размер '{Name}' изменен на {Size}");
    }
}

// ==================== ПРОИЗВОДНЫЙ КЛАСС BOOKS ====================

public class Books : Product
{
    private string _author;
    private int _pages;
    private string _genre;
    private string _language;
    private bool _isBestseller;

    public string Author
    {
        get { return _author; }
        set { _author = !string.IsNullOrWhiteSpace(value) ? value : throw new ArgumentException("Автор не может быть пустым!"); }
    }

    public int Pages
    {
        get { return _pages; }
        set { _pages = value > 0 ? value : throw new ArgumentOutOfRangeException("Страниц должно быть больше нуля!"); }
    }

    public string Genre
    {
        get { return _genre; }
        set { _genre = value ?? "Не указан"; }
    }

    public string Language
    {
        get { return _language; }
        set { _language = value ?? "Русский"; }
    }

    public bool IsBestseller
    {
        get { return _isBestseller; }
        set { _isBestseller = value; }
    }

    public Books(string name, decimal price, string manufacturer, string author, int pages,
                 string category, string country, double rating, int stockQuantity,
                 string genre, string language, bool isBestseller)
        : base(name, price, manufacturer, category, country, rating, stockQuantity)
    {
        Author = author;
        Pages = pages;
        Genre = genre;
        Language = language;
        IsBestseller = isBestseller;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Автор: {Author}, Страниц: {Pages}, Жанр: {Genre}, Язык: {Language}";
    }

    public override decimal Discount()
    {
        decimal rate = Pages > 500 ? 0.8m : 0.9m;
        decimal discountPrice = Price * rate;
        Console.WriteLine($"[Books] Скидка для '{Name}' ({Pages} стр.): {(100 - rate * 100)}%. Цена: {discountPrice:C}");
        return discountPrice;
    }

    public void ShowAuthorInfo()
    {
        Console.WriteLine($"[Books] Книга '{Name}' написана автором: {Author}");
    }

    public string GetGenreInfo()
    {
        return $"Жанр книги '{Name}': {Genre}";
    }

    public void Translate(string newLanguage)
    {
        Language = newLanguage;
        Console.WriteLine($"[Books] Книга '{Name}' переведена на язык: {Language}");
    }

    public void MarkAsBestseller()
    {
        IsBestseller = true;
        Console.WriteLine($"[Books] '{Name}' теперь бестселлер!");
    }
}

// ==================== УПРАВЛЕНИЕ ЗАВИСИМОСТЯМИ (Dependency Injection) ====================

public class InventoryService
{
    private readonly IProductActions _product;
    private readonly List<string> _operations = new List<string>();

    public InventoryService(IProductActions product)
    {
        _product = product;
        _operations.Add($"Инициализирован сервис для продукта: {_product.GetCategory()}");
    }

    public void RestockProduct(int quantity)
    {
        _product.Restock(quantity);
        _operations.Add($"Пополнение на {quantity} шт.");
    }

    public void DisplayProductInfo()
    {
        string category = _product.GetCategory();
        string warranty = _product.GetWarrantyInfo();
        
        Console.WriteLine($"\n--- Информация из сервиса ---");
        Console.WriteLine($"Категория: {category}");
        Console.WriteLine(warranty);
    }

    public void ShowOperationsLog()
    {
        Console.WriteLine("\n=== Журнал операций ===");
        foreach (var op in _operations)
        {
            Console.WriteLine($"• {op}");
        }
    }
}

public class ProductService
{
    private readonly Product _product;

    public ProductService(Product product)
    {
        _product = product;
    }

    public void ApplyDiscount()
    {
        _product.Discount();
    }

    public void DisplayDetails()
    {
        _product.Display();
    }

    public void UpdateRating(double points)
    {
        _product.IncreaseRating(points);
    }
}

// ==================== ОСНОВНАЯ ПРОГРАММА (ДЕМОНСТРАЦИЯ) ====================

Console.WriteLine("=== 1. СОЗДАНИЕ ПРОДУКТОВ С НОВЫМИ АТРИБУТАМИ ===\n");

Product electronics = new Electronics(
    name: "Смартфон Galaxy S23",
    price: 79999m,
    manufacturer: "Samsung",
    warrantyPeriod: 24,
    category: "Электроника",
    country: "Южная Корея",
    rating: 4.8,
    stockQuantity: 15,
    power: 15,
    color: "Черный",
    isWireless: true
);

Product clothing = new Clothing(
    name: "Футболка спортивная",
    price: 2499m,
    manufacturer: "Adidas",
    size: "L",
    category: "Одежда",
    country: "Германия",
    rating: 4.5,
    stockQuantity: 50,
    material: "Полиэстер",
    season: "Лето",
    isWashed: false
);

Product book = new Books(
    name: "Изучаем C#",
    price: 2999m,
    manufacturer: "Питер",
    author: "Герберт Шилдт",
    pages: 1200,
    category: "Книги",
    country: "США",
    rating: 4.7,
    stockQuantity: 20,
    genre: "Техническая",
    language: "Русский",
    isBestseller: true
);

Console.WriteLine("=== 2. ДЕМОНСТРАЦИЯ НОВЫХ МЕТОДОВ ===\n");

Console.WriteLine("[Electronics]");
var electronic = (Electronics)electronics;
electronic.TurnOn();
electronic.ToggleWireless();
Console.WriteLine(electronic.GetPowerInfo());
electronic.UpdatePrice(74999m);
electronic.IncreaseRating(0.1);

Console.WriteLine();

Console.WriteLine("[Clothing]");
var cloth = (Clothing)clothing;
cloth.Wash();
Console.WriteLine(cloth.GetSeasonInfo());
cloth.ChangeSize("XL");

Console.WriteLine();

Console.WriteLine("[Books]");
var bookItem = (Books)book;
bookItem.ShowAuthorInfo();
Console.WriteLine(bookItem.GetGenreInfo());
bookItem.Translate("Английский");
bookItem.MarkAsBestseller();

Console.WriteLine("\n=== 3. ЯВНАЯ РЕАЛИЗАЦИЯ ИНТЕРФЕЙСА ===\n");

IProductActions productActions = electronics;
productActions.Restock(10);
Console.WriteLine($"Категория: {productActions.GetCategory()}");
Console.WriteLine(productActions.GetWarrantyInfo());

Console.WriteLine("\n=== 4. УПРАВЛЕНИЕ ЗАВИСИМОСТЯМИ (Dependency Injection) ===\n");

InventoryService inventoryService = new InventoryService(electronics);
inventoryService.RestockProduct(5);
inventoryService.DisplayProductInfo();
inventoryService.ShowOperationsLog();

Console.WriteLine();

InventoryService bookInventory = new InventoryService(book);
bookInventory.RestockProduct(15);
bookInventory.DisplayProductInfo();

Console.WriteLine("\n=== 5. ПОЛИМОРФИЗМ ЧЕРЕЗ SERVICE ===\n");

ProductService electronicService = new ProductService(electronics);
electronicService.DisplayDetails();
electronicService.ApplyDiscount();
electronicService.UpdateRating(0.1);

Console.WriteLine();

ProductService clothingService = new ProductService(clothing);
clothingService.DisplayDetails();
clothingService.ApplyDiscount();

Console.WriteLine("\n=== 6. ИТОГОВАЯ ИНФОРМАЦИЯ ===\n");

Console.WriteLine(electronics.GetInfo());
Console.WriteLine(clothing.GetInfo());
Console.WriteLine(book.GetInfo());

Console.WriteLine("\n=== ПРОГРАММА ЗАВЕРШЕНА ===");
```
